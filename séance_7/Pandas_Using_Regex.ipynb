# Pandas String Methods

This notebook is a **conceptual and practical reference** for two essential Pandas string methods:

- `.str.contains()` → filtering text data using patterns
- `.str.replace()` → cleaning and transforming text data

The goal is **understanding usage**, not memorizing regex syntax.


## Why Pandas String Methods?

In real datasets, text columns often contain:
- currency symbols
- extra spaces
- inconsistent wording
- numbers embedded in text

Pandas provides the `.str` accessor to apply **string operations column-wise**.

```python
import pandas as pd

df = pd.DataFrame({
    'text': [
        'In stock (22 available)',
        'Out of stock',
        'In stock (5 available)',
        'Only 1 left in stock'
    ]
})

df
```

## Filtering Rows

`.str.contains()` checks whether a **pattern exists** in each value of a column.

**Typical use cases:**
- keep rows containing a keyword
- exclude rows based on text
- filter data before analysis


```python
df.head()
```

```python
df[df['text'].str.contains('In stock')]
```

```python
# Keep rows containing 'In stock'
df[df['text'].str.lower().str.contains('in stock')]
```

### Case sensitivity

By default, `.str.contains()` is **case-sensitive**.

```python
df[df['text'].str.contains('in stock', case=False)]
```

### Handling missing values

If a column contains `NaN`, use `na=False` to avoid errors.

```python
df[df['text'].str.contains(r'\(.*\)', na=False, regex=True)]
```

```python
df[df['text'].str.contains('stock', na=False)]
```

## Cleaning Text

`.str.replace()` replaces parts of strings using **literal text or regex patterns**.

**Typical use cases:**
- remove currency symbols
- delete unwanted characters
- standardize text formats


```python
# Remove parentheses and their content
df['clean_text'] = df['text'].str.replace(r'\(.*\)', '', regex=True)

df['clean_text_2'] = df['text'].str.replace('In stock', 'in stock', regex=False)

df
```

### Replacing literal text

Regex is optional — for simple replacements it can be disabled.

```python
df["number"] = "22"
```

```python
df.info()
```

```python
df['standardized'] = df['text'].str.replace('Out of stock', 'Unavailable', regex=False)
df
```

## Common Regex Patterns

Some **very common patterns** you will see in practice:

- `\d+` → one or more digits
- `\(.*\)` → text inside parentheses
- `[^0-9]` → anything that is NOT a digit


```python
df["text"].str
```

```python
df["text"].info()
```

```python
# Extract numbers from text
df['stock_number'] = df['text'].str.replace(r'[^0-9]', '', regex=True)
df
```

## Summary

| Method | Purpose |
|------|--------|
| `.str.contains()` | Filter rows based on text |
| `.str.replace()` | Modify or clean text values |

**Rule of thumb:**
- Need to **keep or discard rows** → `contains`
- Need to **change text values** → `replace`
