# Basic Data Visualization

- Histograms → distribution of a numerical variable
- Box plots → spread, median, outliers
- Bar charts → comparison between categories

The goal is to **choose the right chart and build it correctly**.


## Setup & Sample Data


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'price': [12, 15, 18, 22, 25, 30, 35, 40, 45, 50],
    'category': ['A', 'A', 'B', 'B', 'B', 'C', 'C', 'C', 'C', 'C']
})

df
```

## Histograms — Distribution of a Numerical Variable

A **histogram** shows how values are distributed across ranges.

**Typical questions answered:**
- Are values concentrated or spread out?
- Is the distribution skewed?

**Use when:**
- You have **one numerical column**

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df['price'])
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Histogram of Prices')
plt.show()
```

### Common parameters

- `bins` → number of bars
- `edgecolor` → separates bars visually (optional)

```python
plt.hist(df['price'], bins=5)
plt.show()
```

## Box Plots — Spread & Outliers

A **box plot** summarizes a numerical variable using:
- minimum
- first quartile (Q1)
- median
- third quartile (Q3)
- maximum / outliers

**Use when:**
- You want a **statistical summary**

```python
plt.boxplot(df['price'])
plt.ylabel('Price')
plt.title('Box Plot of Prices')
# plt.show()
# ;
```

```python
plt.boxplot(df['price'])
plt.ylabel('Price')
plt.title('Box Plot of Prices')
plt.show()
# ;
```

### Box plot by category

Useful to compare distributions between groups.

```python
df.head()
```

```python
df.boxplot(column='price', by='category')
plt.title('Price Distribution by Category')
plt.suptitle('')
plt.show()
```

## Bar Charts — Comparing Categories

A **bar chart** compares values across categories.

**Typical questions answered:**
- Which category has the highest value?
- How do categories compare?

**Use when:**
- You have **categorical data**

### Step 1 — Aggregate the data

Bar charts usually require **grouping** first.

```python
avg_price = df.groupby('category')['price'].mean()
avg_price = df.groupby('category').agg(avg_price=('price', 'mean')).reset_index()
avg_price
```

### Step 2 — Plot the bar chart

```python
avg_price.plot(kind='bar')
plt.xlabel('Category')
plt.ylabel('Average Price')
plt.title('Average Price per Category')
plt.show()
```

## Choosing the Right Chart — Summary

| Question | Chart |
|--------|-------|
| Distribution of values | Histogram |
| Spread & outliers | Box plot |
| Compare categories | Bar chart |
